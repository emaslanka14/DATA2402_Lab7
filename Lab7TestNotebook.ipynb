```python
import numpy as np 
import lab7_2 as lab
print("Lab 7 Testing Notebook")
```

```python
data = np.genfromtxt('stock_prices_intro.csv', delimiter=',', names=True, dtype=None, encoding='utf-8')

dates = data['Date']

openp = data['Open']

closep = data['Close']

vol = data['Volume']

print("Number of records:", len(data))
print("Column Names:", data.dtype.names)
```

```python
assert len(data) == 20
assert data.dtype.names == ('Date', 'Open', 'Close', 'Volume')
assert len(dates) == 20
assert len(openp) == 20
assert len(closep) == 20
assert len(vol) == 20
assert np.issubdtype(openp.dtype, np.number)
assert np.issubdtype(closep.dtype, np.number)
assert np.issubdtype(vol.dtype, np.number)
print("PASSED: CSV loading and array extraction")
```

```python
rate = lab.task2(closep, openp, dates)
expected_rate = (closep - openp) / openp * 100
assert np.allclose(rate, expected_rate)
assert len(rate) == 20
print("Expected first five: ", expected_rate[:5])
print("Actual first five: ", rate[:5])
print("PASSED: Daily return calculation")
```

```python
mean_rate, median_rate, std_rate = lab.task3(rate)
assert np.isclose(mean_rate, np.mean(rate))
assert np.isclose(median_rate, np.median(rate))
assert np.isclose(std_rate, np.std(rate))
print("PASSED: Mean, median, and standard deviation")
```

```python
bestIndex, worstIndex = lab.task4(dates, openp, closep, rate)
assert bestIndex == np.argmax(rate)
assert worstIndex == np.argmin(rate)
assert rate[bestIndex] == np.max(rate)
assert rate[worstIndex] == np.min(rate)
print("PASSED: Best and worst trading days")
```

```python
upDays, downDays = lab.task5(rate)
expectedUpDays = np.sum(rate > 0)
expectedDownDays = np.sum(rate < 0)

assert upDays == expectedUpDays
assert downDays == expectedDownDays
assert upDays + downDays + np.sum(rate == 0) == 20
print("PASSED: Positive and negative trading days")
```

```python
cum_return = lab.task6(rate)
expectedReturn = (np.prod(1 + rate / 100) - 1) * 100
assert np.isclose(cum_return, expectedReturn)
print("Expected:", expectedReturn)
print("Actual:", cum_return)
print("PASSED: Cumulative return calculation")
```
